```python
# ================================================
# Tic Tac Toe with Minimax and Alpha-Beta Pruning
# ================================================

import math
import random

# ---------- Board utilities ----------
def print_board(board):
    """Print the Tic Tac Toe board in a 3x3 grid."""
    print("   0   1   2")
    for i, row in enumerate(board):
        print(i, " | ".join(row))
        if i < 2:
            print("  ---+---+---")
    print()

def is_moves_left(board):
    """Check if there are moves left on the board."""
    for row in board:
        if " " in row:
            return True
    return False

def evaluate(board):
    """Evaluate the board state: +10 for 'O' win, -10 for 'X' win, 0 otherwise."""
    # Check rows
    for row in board:
        if row[0] == row[1] == row[2] != " ":
            return 10 if row[0] == "O" else -10

    # Check columns
    for col in range(3):
        if board[0][col] == board[1][col] == board[2][col] != " ":
            return 10 if board[0][col] == "O" else -10

    # Check diagonals
    if board[0][0] == board[1][1] == board[2][2] != " ":
        return 10 if board[0][0] == "O" else -10
    if board[0][2] == board[1][1] == board[2][0] != " ":
        return 10 if board[0][2] == "O" else -10

    return 0

# ---------- Minimax with Alpha-Beta ----------
def minimax(board, depth, is_max, alpha, beta, max_depth):
    """Minimax algorithm with alpha-beta pruning and depth limit."""
    score = evaluate(board)

    # Terminal states
    if score == 10 or score == -10:
        return score
    if not is_moves_left(board) or depth == max_depth:
        return 0

    if is_max:  # Maximizer -> 'O' (AI)
        best = -math.inf
        for i in range(3):
            for j in range(3):
                if board[i][j] == " ":
                    board[i][j] = "O"
                    best = max(best, minimax(board, depth + 1, False, alpha, beta, max_depth))
                    board[i][j] = " "
                    alpha = max(alpha, best)
                    if beta <= alpha:
                        return best
        return best
    else:  # Minimizer -> 'X' (Opponent)
        best = math.inf
        for i in range(3):
            for j in range(3):
                if board[i][j] == " ":
                    board[i][j] = "X"
                    best = min(best, minimax(board, depth + 1, True, alpha, beta, max_depth))
                    board[i][j] = " "
                    beta = min(beta, best)
                    if beta <= alpha:
                        return best
        return best

# ---------- AI Move ----------
def find_best_move(board, player, max_depth=9):
    """Find the best move for the AI ('O' or 'X')."""
    best_val = -math.inf if player == "O" else math.inf
    best_move = (-1, -1)

    for i in range(3):
        for j in range(3):
            if board[i][j] == " ":
                board[i][j] = player
                move_val = minimax(board, 0, player == "X", -math.inf, math.inf, max_depth)
                board[i][j] = " "

                if player == "O" and move_val > best_val:
                    best_val = move_val
                    best_move = (i, j)
                elif player == "X" and move_val < best_val:
                    best_val = move_val
                    best_move = (i, j)

    return best_move

# ---------- Game Loop ----------
def play_game(mode="human-vs-computer", depth_x=3, depth_o=3):
    """Play Tic Tac Toe game in console."""
    board = [[" " for _ in range(3)] for _ in range(3)]
    print_board(board)

    current_player = "X"  # Human always starts as "X"

    while True:
        if current_player == "X":
            if mode == "human-vs-computer":
                row, col = map(int, input("Enter your move (row col): ").split())
            else:  # Computer vs Computer
                row, col = find_best_move(board, "X", depth_x)
                print(f"Computer X chooses: {row}, {col}")
        else:  # Player O
            row, col = find_best_move(board, "O", depth_o)
            print(f"Computer O chooses: {row}, {col}")

        if board[row][col] == " ":
            board[row][col] = current_player
        else:
            print("Invalid move! Try again.")
            continue

        print_board(board)

        score = evaluate(board)
        if score == 10:
            print("O wins!")
            break
        elif score == -10:
            print("X wins!")
            break
        elif not is_moves_left(board):
            print("It's a draw!")
            break

        current_player = "O" if current_player == "X" else "X"

# ---------------- RUN GAME ----------------
# Mode 1: Human vs Computer
# play_game("human-vs-computer", depth_x=3, depth_o=5)

# Mode 2: Computer vs Computer
play_game("computer-vs-computer", depth_x=3, depth_o=5)
```
